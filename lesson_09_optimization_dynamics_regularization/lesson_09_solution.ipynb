## Full worked solution
This version contains one possible complete implementation. Use it after attempting the exercise notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(9)
x = rng.normal(size=120)
y = 2.8*x + rng.normal(0, 0.4, size=x.shape)

def loss_and_grad(w):
    pred = w*x
    err = pred - y
    loss = np.mean(err**2)
    grad = 2*np.mean(err*x)
    return loss, grad

def run_gd(eta, steps=60):
    w = 0.0
    losses = []
    for _ in range(steps):
        loss, grad = loss_and_grad(w)
        losses.append(loss)
        w -= eta*grad
    return w, np.array(losses)

plt.figure(figsize=(7, 4))
for eta in [0.01, 0.1, 0.8]:
    w, losses = run_gd(eta)
    plt.plot(losses, label=f"eta={eta}, final w={w:.2f}")
plt.yscale("log")
plt.xlabel("step")
plt.ylabel("MSE")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

def run_adam(alpha=0.1, steps=60, beta1=0.9, beta2=0.999, eps=1e-8):
    w, m, v = 0.0, 0.0, 0.0
    losses = []
    for t in range(1, steps+1):
        loss, g = loss_and_grad(w)
        losses.append(loss)
        m = beta1*m + (1-beta1)*g
        v = beta2*v + (1-beta2)*g*g
        mhat = m/(1-beta1**t)
        vhat = v/(1-beta2**t)
        w -= alpha*mhat/(np.sqrt(vhat)+eps)
    return w, np.array(losses)

w_adam, losses_adam = run_adam()
print(f"Adam final w = {w_adam:.3f}")
print("Regularization and early stopping are control mechanisms: they stop optimization from chasing noise.")
```
