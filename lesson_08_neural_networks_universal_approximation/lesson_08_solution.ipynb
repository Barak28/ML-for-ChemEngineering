```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

z = np.linspace(-5, 5, 400)
plt.figure(figsize=(7, 4))
plt.plot(z, 1/(1+np.exp(-z)), label="sigmoid")
plt.plot(z, np.tanh(z), label="tanh")
plt.plot(z, np.maximum(0, z), label="ReLU")
plt.xlabel("z")
plt.ylabel("activation")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

rng = np.random.default_rng(8)
X = rng.uniform(-3, 3, size=(90, 1))
y = np.sin(2*X[:,0]) + 0.25*X[:,0]**2 + rng.normal(0, 0.12, 90)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

models = {
    "small": make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(8,), max_iter=4000, random_state=1)),
    "large": make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(80,80), alpha=1e-4, max_iter=4000, random_state=1)),
}
grid = np.linspace(-3.3, 3.3, 300).reshape(-1, 1)
plt.figure(figsize=(8, 4))
plt.scatter(X_train[:,0], y_train, s=20, label="train", alpha=0.6)
for name, model in models.items():
    model.fit(X_train, y_train)
    train_rmse = mean_squared_error(y_train, model.predict(X_train), squared=False)
    test_rmse = mean_squared_error(y_test, model.predict(X_test), squared=False)
    print(f"{name:5s} train RMSE={train_rmse:.3f}, test RMSE={test_rmse:.3f}")
    plt.plot(grid[:,0], model.predict(grid), label=name)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print("For PINNs and derivative-based physics, smooth activations such as tanh are often safer than ReLU.")
```
