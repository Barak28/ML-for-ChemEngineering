```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, accuracy_score, ConfusionMatrixDisplay

rng = np.random.default_rng(5)
n = 260

# Synthetic cross-disciplinary engineering system:
# a compact motor-pump module cooled by a circulating fluid.
coolant_flow = rng.uniform(0.7, 5.5, n)          # L/min
heat_load = rng.uniform(35, 95, n)              # W
ambient_temp = rng.uniform(20, 38, n)           # deg C
vibration = rng.gamma(shape=2.0, scale=0.35, size=n)  # arbitrary health indicator

# Physically interpretable temperature model:
# low flow creates a threshold-like cooling collapse.
thermal_resistance = 1.25 / np.maximum(coolant_flow, 0.35)
low_flow_penalty = np.where(coolant_flow < 2.7, 12.0 + 4.0 * (2.7 - coolant_flow), 0.0)
T_max = (
    42
    + thermal_resistance * heat_load
    + 0.85 * (ambient_temp - 25)
    + 2.2 * vibration
    + low_flow_penalty
    + rng.normal(0, 2.6, n)
)

overheat = (T_max > 85).astype(int)

X = pd.DataFrame({
    "coolant_flow_L_min": coolant_flow,
    "heat_load_W": heat_load,
    "ambient_temp_C": ambient_temp,
    "vibration_index": vibration,
})
y_reg = T_max
y_cls = overheat

X_train, X_test, y_train, y_test, yc_train, yc_test = train_test_split(
    X, y_reg, y_cls, test_size=0.25, random_state=2, stratify=y_cls
)

print(X.head())
print("Overheating fraction:", y_cls.mean().round(3))
```
