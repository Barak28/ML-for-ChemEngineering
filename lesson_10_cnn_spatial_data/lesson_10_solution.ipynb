## Full worked solution
This version contains one possible complete implementation. Use it after attempting the exercise notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(10)
img = rng.normal(0, 0.05, size=(40, 70))
img[12:28, 28:45] += 1.0

Kx = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
Ky = Kx.T

def conv2d_same(image, kernel):
    kh, kw = kernel.shape
    ph, pw = kh//2, kw//2
    padded = np.pad(image, ((ph, ph), (pw, pw)), mode="edge")
    out = np.zeros_like(image, dtype=float)
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            patch = padded[i:i+kh, j:j+kw]
            out[i, j] = np.sum(patch * kernel)
    return out

gx = conv2d_same(img, Kx)
gy = conv2d_same(img, Ky)
edge = np.sqrt(gx**2 + gy**2)

fig, ax = plt.subplots(1, 3, figsize=(12, 3.5))
for a, data, title in zip(ax, [img, gx, edge], ["Raw spatial signal", "Vertical edge response", "Edge magnitude"]):
    im = a.imshow(data, cmap="viridis")
    a.set_title(title)
    a.axis("off")
    plt.colorbar(im, ax=a, fraction=0.046)
plt.tight_layout()
plt.show()
print("The same small filter is reused at every location: this is parameter sharing.")
```
