## Full worked solution
This version contains one possible complete implementation. Use it after attempting the exercise notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

tokens = ["C", "C", "=O", "O"]
rng = np.random.default_rng(12)
X = rng.normal(size=(len(tokens), 6))
Wq = rng.normal(scale=0.5, size=(6, 4))
Wk = rng.normal(scale=0.5, size=(6, 4))
Wv = rng.normal(scale=0.5, size=(6, 4))

Q, K, V = X @ Wq, X @ Wk, X @ Wv

def softmax(a, axis=-1):
    a = a - np.max(a, axis=axis, keepdims=True)
    e = np.exp(a)
    return e / e.sum(axis=axis, keepdims=True)

scores = Q @ K.T / np.sqrt(Q.shape[1])
A = softmax(scores, axis=1)
context = A @ V

plt.figure(figsize=(4.8, 4))
plt.imshow(A, cmap="viridis", vmin=0, vmax=1)
plt.xticks(range(len(tokens)), tokens)
plt.yticks(range(len(tokens)), tokens)
plt.colorbar(label="attention weight")
plt.title("Scaled dot-product attention")
plt.show()
print("Context vectors shape:", context.shape)
print("Attention shows learned relevance inside the model; it is not, by itself, proof of causality.")
```
