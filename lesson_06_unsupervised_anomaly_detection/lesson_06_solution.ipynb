## Full worked solution
This version contains one possible complete implementation. Use it after attempting the exercise notebook.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(6)
nominal = rng.normal([2, 8], [0.45, 0.35], size=(90, 2))
drift = np.column_stack([np.linspace(3.5, 8.5, 24), np.linspace(7.0, 2.0, 24)]) + rng.normal(0, 0.15, (24, 2))
X = np.vstack([nominal, drift])
Xs = StandardScaler().fit_transform(X)

kmeans = KMeans(n_clusters=2, n_init=20, random_state=0).fit(Xs)
db = DBSCAN(eps=0.55, min_samples=6).fit(Xs)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].scatter(X[:,0], X[:,1], c=kmeans.labels_, cmap="tab10", s=28)
ax[0].set_title("k-means: forced spherical clusters")
ax[1].scatter(X[:,0], X[:,1], c=db.labels_, cmap="tab10", s=28)
ax[1].set_title("DBSCAN: dense cluster + noise")
for a in ax:
    a.set_xlabel("Latent feature 1")
    a.set_ylabel("Latent feature 2")
    a.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print("DBSCAN labels:", sorted(set(db.labels_)))
print("Noise points are labeled -1.")
print("DBSCAN is useful here because drift is a sparse trajectory, not a compact spherical cluster.")
```
